```python
import matplotlib as mpl
x = ['2019', '2020', '2021']
y = [-100, -200, -300]

mpl.rcParams['axes.unicode_minus'] = False

plt.bar(x, y)
plt.show()
```

```python
# x축 겹침 문제
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.arange(2005, 2026)
y = np.arange(len(x))

plt.figure(figsize=(8,3))

plt.bar(x, y)
plt.xticks(x)

plt.xticks(rotation = 45, ha='right') # 겹침 문제 해결 방법
plt.show()
```
