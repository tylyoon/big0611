```python
import pandas as pd
import numpy as np
```

```python
#딕셔너리로 생성
product_dict = {
    'product_name': ['노트북', '마우스', '키보드', '모니터'],
    'price': [1200000, 25000, 80000, 350000],
    'stock': [10, 50, 30, 15],
    'category': ['전자제품', '주변기기', '주변기기', '전자제품']
}
```

```python
## 데이타 프레임 복사
```

```python
# 원본 데이터프레임 생성
original_df = pd.DataFrame({
    'name': ['김철수', '이영희', '박민수'],
    'age': [25, 30, 35],
    'salary': [3000, 4000, 5000]
})

print("=== 원본 데이터프레임 ===")
print(original_df)
```

```python
# 깊은 복사 vs 참조 복사
# 깊은 복사 : df.copy() - 원본이 바뀌지 않는다.
deep_copy_df = original_df.copy() # 원본이 바뀌지 않는다.

reference_df = original_df

print(deep_copy_df)
print(reference_df)
```

```python
# 특정 아이템을 선택
# deep_copy_df.loc[행,열]
deep_copy_df.loc[0,'salary'] = 3500
print(original_df)
reference_df.loc[1,'age'] = 31
print(deep_copy_df)
print(original_df)
print(reference_df)

```

```python
# numpy copy
arr_org = np.arange(10).reshape(2,5)
arr_copy = arr_org.copy()
print(arr_org)
print(arr_copy)
```

```python
# 딕셔너리 복사
dict_org = {
    'group1' : {'name' : '김철수', 'age':25}
}

print(dict_org)
dict_copy = dict_org.copy()
print(dict_copy)
dict_copy['group1']['age'] = 30
print(dict_copy)
print(dict_org)


# .copy() : 얕은복사, copy.deepcopy() : 깊은복사
```

```python
# 깊은복사
import copy


dict_org1 = {
    'group1' : {'name' : '김철수', 'age':25}
}

print(dict_org1)
dict_copy1 = copy.deepcopy(dict_org1)
print(dict_copy1)
dict_copy1['group1']['age'] = 30
print(dict_copy1)
print(dict_org1)
```

## 행과 열 삭제


```python
test_data = pd.DataFrame({
    'id': [1, 2, 3, 4, 5, 6],
    'name': ['김철수', '이영희', '박민수', '김철수', '정수진', '이영희'],
    'age': [25, 30, None, 25, 28, 30],
    'city': ['서울', '부산', '대구', '서울', None, '부산'],
    'salary': [3000, 4000, 5000, 3000, 3500, 4000],
    'temp_col': ['임시', '임시', '임시', '임시', '임시', '임시']
})
print("=== 원본 데이터 ===")
print(test_data)
```

```python
# 특정 열 삭제
# df.drop()
# axis : 삭제출,(0=행, 1=열)
df_drop_col = test_data.drop('temp_col',axis = 1)
print(' ======== 열삭제 후 =======')

df_drop_cols = test_data.drop(['city','age'],axis = 1)
print(df_drop_col)
print(test_data)
```

```python
# 특정 행 삭제(인덱스 기준) : axis = 0
df_drop_row = test_data.drop([0,5],axis = 0)
print(df_drop_row)
```

```python
# 중복행 삭제 (keep = 'fist':첫번째(기본값), keep = 'last':마지막,k=False: 중복된 모두)
# duplicates = test_data.drop_duplicates(subset=[컬럼명,'age'],keep= 'False')
df_no_duplicates = test_data.drop_duplicates(subset=['name','age'],keep= False)
print(df_no_duplicates)
```

```python
# 결측치 삭제(NaN:not a number)
print(test_data)
df_no_na = test_data.dropna()
print(df_no_na)
```

```python
#특정 columne의 결측치 삭제
print(test_data)
df_no_na_age = test_data.dropna(subset=['age'])

print(df_no_na_age)
```
