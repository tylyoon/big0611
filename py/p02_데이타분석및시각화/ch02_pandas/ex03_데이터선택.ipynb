```python

import pandas as pd
```

```python
# 예시 데이터 생성
sales_data = pd.DataFrame({
    'product': ['노트북', '마우스', '키보드', '모니터', '스피커'],
    'price': [1200000, 25000, 80000, 350000, 120000],
    'quantity': [10, 50, 30, 15, 25],
    'category': ['전자제품', '주변기기', '주변기기', '전자제품', '주변기기']
})
print(sales_data)
```

```python
product_series = sales_data['product']
print(product_series)
```

```python
# 다중 컬럼 선택
price_quantity = sales_data[['product', 'price', 'quantity']]
print(price_quantity)
```

```python
# 행 슬라이싱 (첫 3개 행 선택)
#first_three = sales_data[0:3] : 인덱스 번호
first_three = sales_data[0:3]
print(first_three)
```

```python
## 불리언 인덱싱(Boolean Indexing)
# 가격이 10만원 이상인 제품 선택 : 가격 >= 100000
expensive_products = sales_data[sales_data['price'] >= 100000]
print(expensive_products)
```

```python
#2. df.query()
#high_quantity = sales_data[sales_data['quantity'] > 20]
high_quantity = sales_data.query('quantity > 20')
print(high_quantity)
```

```python
# loc와 iloc을 활용한 라벨/위치 기반 선택
print(sales_data)
sales_data_indexed = sales_data.set_index('product')
print(sales_data_indexed)
```

```python
# df.loc[행,열] : 라벨기반 선택
notebook_info = sales_data_indexed.loc['노트북', 'price']
print(f'노트북 가격:{notebook_info:,}')
```

```python
# 여러 제품의 가격과 수량 선택
selected_data = sales_data_indexed.loc[['노트북', '모니터'], ['price', 'quantity']]
print(selected_data)
```

```python
# 첫번째와 세번째 행,두번째와 세번째 컬럼
#df.iloc[행index,얄index]
print(sales_data)
positional_data = sales_data.iloc[1, 2]
print(positional_data)
positional_data1 = sales_data.iloc[[0, 1, 2], [1, 2]]
print(positional_data1)
positional_data2 = sales_data.iloc[0:3, 1:3]
print(positional_data2)
```

```python
## 다중 조건 선택
# 가격이 50000원 이상이면서 (and,&),수량이 20개 이상인 제품
# 가격 >= 50000 & 수량 >= 20
# sales_data['price'] >= 50000 & sales_data['quantity] >= 20
filtered_products = sales_data[(sales_data['price'] >= 50000) & (sales_data['quantity'] >= 20)]
print(sales_data)
print(end='\n')
print(filtered_products)
```

```python
# 전자제품이거나 가격이 30000원 이하인 제품
electronic_or_cheap = sales_data[(sales_data['category'] == '전자제품') | (sales_data['price'] <= 30000)]
print(electronic_or_cheap)
```

```python
## 그룹별 데이타 선택
```

```python
# 카테고리별 그룹화 후 전자제품그룹 맡
#df.groupby('컬럼명').get_ground('데이터')
#get_groud
print(sales_data)
electrics_group = sales_data.groupby('category').get_group('전자제품')
print(electrics_group)
```

```python
# 각 카테고리별 평균 가격을 원본 데이터에 추가해라
sales_data['category_avg_price'] = sales_data.groupby('category')['price'].transform('mean')
print(sales_data[['product', 'category', 'price', 'category_avg_price']])
```

```python
# 시계열 데이터 생성
dates = pd.date_range('2026-01-01', periods=5, freq='D')
time_series_data = pd.DataFrame({
    'date': dates,
    'sales': [100, 150, 200, 120, 180],
    'returns': [5, 8, 12, 6, 9]
})
time_series_data.set_index('date', inplace=True)

print(time_series_data)
```

```python
#특정날짜 범위 선택
jan_1_to_3 = time_series_data.loc['2026-01-01':'2026-01-03']
print(jan_1_to_3)
```

```python
# 연도기준 필터링
# 2026 데이타만 표시
year_2024 = time_series_data[time_series_data.index.year == 2026]
print(year_2024)
```
