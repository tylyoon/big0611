```python
import pandas as pd
import numpy as np
```

```python
# Seriese 생성
#pd.Series() 
temperatures = pd.Series(
    [23.5,25.1,19.8,30.2],
    index=['월','화','수','목'],
    name = '온도'
)
print(temperatures)
```

```python
# DataFrame 예시 : 2차원 data
weather_data = pd.DataFrame({
    '온도' : [23.5,25.1,19.8,30.2],
    '습도' : [65,70,60,75],
    '강수량' : [0,5,10,20]
}, index=['월','화','수','목'])

print(weather_data)
```

```python
# Num data
# 숫자 인텍스만 사용 가능
data_array = np.array([
    [23.5,69, 0],
    [25.1,65,0],
    [19.8, 70, 5.2],
    [30.2,55,0]])
print(data_array)
print(data_array[0][0])
```

```python
# 의미 있는 인덱스와 컬러명을 사용한다.
# .loc[]
print(weather_data['온도'])
print(weather_data.loc['수', '온도'])
```
