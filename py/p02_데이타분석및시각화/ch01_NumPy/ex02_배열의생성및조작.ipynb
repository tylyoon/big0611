```python
## 배열의 저장과 불러오기
```

```python

# 배열 저장하기 = np.save(파일이름, 배열이름) : .npy 형식으로 저장
# np.savez(file,array1,array2,array3....): 여러배열을 .mpz로 저장
# np.savetxt(fname,array,fmt='%d',delimiter=','):텍스트 파이로 저장
# np.savez_compressed(file,array1,array2,array3....): 압축하여 저장
```

```python
# 매서트 체이닝 : 모듈.함수().메서드()
arr = np.arange(10)
print(arr)
arr.reshape(2, 5)
arr = np.arange(10).reshape(2, 5)
print(arr)
```

```python
#배열 저장하기
np.save('array.npy', arr)
```

```python
#배열 불러오기 - np.load()
# np.save()
# np.load('파일명')
load_arr = np.load('array.npy')
print(load_arr)
```

```python
#여러 배열 생성
arr1 = np.arange(10)
arr2 = np.arange(20).reshape(4,5)
arr3 = np.random.rand(3,3)
print(arr3)
```

```python
np.savez('multiful_array',first_array=arr1,second_array=arr2,third_array=arr3)
#np.savez('multiful_array.npz', arr1,arr2,arr3)
```

```python
loaded_data = np.load('multiful_array.npz')
print(loaded_data.files)
```

```python
# 개별 배열 접근
arr1_loaded = loaded_data['first_array']
arr2_loaded = loaded_data['second_array']
arr3_loaded = loaded_data['third_array']

```

```python
# 사용 후 파일 닫기
loaded_data.close()
```

```python
# 압축 저장
np.savez_compressed('compressed_arrays', data1=arr1, data2=arr2, data3=arr3)
```
