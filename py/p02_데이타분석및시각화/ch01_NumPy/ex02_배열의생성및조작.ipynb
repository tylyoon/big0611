# 배열의 생성 및 조작

```python
import numpy as np
```

```python
list1 = [1, 2, 3]
arr1 = np.array(list1)
arr2 = np.array([1, 2, 3])
print(list1)
print(arr1)
print(arr2)
```

```python
#np.arange()
arr1 = np.arange(10)
print(arr1)
arr2 = np.arange(0,10,1)
print(arr2)
arr2 = np.arange(10,0,-1)
print(arr2)
```
