### 브리언 인덱싱
## 블리언 인텍싱의 개념

1. 원본 배열과 동일한 사이즈의 불리언 배열을 생성
2. 조건을 만족하는 인덱스에는 True,아니면 False
3. 이 불리언 배열을 인덱스로 사용하면, True위치의 값을 반환한다.


```python
import numpy as np
```

```python
scores = [80,90,70,65,85,95,90,80,75,80]
score_arr = np.array(scores)
idx =[1,2,3,4]
print(score_arr[idx]) 
```

```python
# 90점 이상인지 확인하는 불리언 배열을 생성
score_arr >=50
```

```python
# 90점 이상만 선택
score_arr[score_arr >= 90]
```

```python
score_arr[score_arr >= 90] = 999
score_arr
```

```python
# 팬시 인덱싱(fancy indexing)
arr = np.arange(20)
indices = [1,3,5,7]
print(arr[indices])
print(score_arr[indices])
```

```python
arr2d = np.arange(12).reshape(3,4)
print(arr2d)
print(arr2d[[0,2],[1,3]]) # arr2d(행,열)의 값을 반환
```

```python
#넘파이 놀리연산자 : and or not
& : 그리고, | : 또는, ~ : not
```

```python
scores = np.array([80,90,70,65,85,95,90,80,75,80])
print(scores[(scores >= 80) & (scores < 90)])
```

```python
# 70점 미만 또는 95점 이상
# 배열[(조건식1) | (조건식2)]
#두 조건식 중 하나 이상 True일 경우 출력
print(scores[(scores < 70) | (scores >= 95)])
```

```python
# 조건에 따른 값 치환
# np.where(조건시,참,거짓)
scores
```

```python
# 90점 이상이면 'A', 아니면 'B'
grades = np.where(scores >= 90, 'A','B')
print(grades)
```

```python
# 실무 활용 예시
# 데이터 필터링
# 실제 데이터 분석에서 자주 사용되는 패턴
temperatures = np.array([22.5, 25.1, 19.8, 30.2, 28.7, 15.3, 32.1])
temperatures
```

```python
# 25도 이상인 날씨만 선텍
hot_days = temperatures[temperatures >= 25]
print(f'25도 이상인 날 : {hot_days}')
```

```python
# 20도 미만인 날씨를 20도로 조정
temperatures[temperatures < 20.0] = 20
print(f'조정된 온도: {temperatures}')
```

```python
### 2차원 데이터 처리
# 성적표 데이터(학생 * 과목)
grades = np.array([
    [85, 92, 78, 90],  # 학생1
    [76, 88, 82, 79],  # 학생2
    [92, 95, 89, 94],  # 학생3
    [68, 72, 75, 71]   # 학생4
])
```

```python
# 각 학생의 평균 점수가 80점 이상인 학생들
# axis=1은 row(행) 방향으로 계산
# axis =0 은 column(열) 방향으로 계산
student_avg = grades.mean(axis=1)
print(student_avg)
```

```python
#불리언 이덱싱
#배열[조건]
good_students = grades[student_avg >= 80]
print(f'평균 80점 이상 학생들의 성적 : {good_students}')
```

```python
# 모든 과목에서 85점 이상인 점수만 추출한다.
excellent_scores = grades[grades >= 85]
print(f'85점 이상 점수들 : {excellent_scores}')
```
