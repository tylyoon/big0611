```python
# 참고 : from numpy import *
import numpy as np
```

```python
#리스트 연산 - 반복문 필요 (리스트 항목에 2를 곱하기)
list_data = [1, 2, 3, 4]
result_list = []
for item in list_data:
    result_list.append(item * 2)
print(result_list)
```

```python
# 배열 연산 - 벡터화 연산
array_data = np.array([1, 2, 3, 4])
result_array = array_data * 2
print(result_array)
```

```python
# np.array() 함수
list1 = [1, 2, 3]
arr1 = np.array(list1)
print(list1)
print(arr1)
```

```python
import time
```

```python

# 리스트 연산
start_time = time.time()
python_list = list(range(1000000))
result_list = [x * 2 for x in python_list]
list_time = time.time() - start_time
print(f"리스트 연산 시간: {list_time:.4f}초")
```

```python
# 배열 연산
start_time = time.time()
numpy_array = np.arange(1000000)
result_array = numpy_array * 2
numpy_time = time.time() - start_time
print(f"NumPy 배열 연산 시간: {numpy_time:.4f}초")
print(f"속도 개선: {list_time / numpy_time:.1f}배")
```

```python
# list를 사용한 연산
start_time = time.time()
# 기준시산이 1970년 1월 1일 00:00:00 (UTC)
print(start_time)
# [0,1,2,,,,,1000000]
python_list = list(range(1000000))
result_list = [x *2 for x in python_list]
# print(result_list)
list_time = time.time() - start_time
print(list_time)
```

```python
#배열을 사용한 연산
start_time = time.time()
print(start_time)
numpy_array = np.arange(1000000)
result_array = numpy_array * 2
print(result_array)
numpy_time = time.time() - start_time
print(f"NumPy 배열 연산 시간: {numpy_time:.6f}초")
```
